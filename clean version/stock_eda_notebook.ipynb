# Stock Dataset EDA Notebook
Exploratory Data Analysis for  stock-indicator dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df=pd.read_csv('csvfiles/NIFTY_50_COMPANIES.csv')
```

```python
# Basic Information
df.info()
display(df.describe())
print(df.isnull().sum())
print('Duplicates:', df.duplicated().sum())
```

```python
# Distribution of Numerical Features
for col in df.select_dtypes(include='number').columns:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(col)
    plt.show()
```

```python
# Boxplots
for col in df.select_dtypes(include='number').columns:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(15,10))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
# Pairplot
cols = ['Close','Volume','RSI_14','MACD','Daily_Return_%']
sns.pairplot(df[cols])
plt.show()
```

```python
# Company Distribution (edit column name if needed)
company_counts = df['Ticker'].value_counts()
display(company_counts)
company_counts.plot(kind='bar', figsize=(15,5))
plt.show()
```

```python
# Company-wise averages
display(df.groupby('Ticker')['RSI_14'].mean().sort_values())
display(df.groupby('Ticker')['Volume'].mean())
display(df.groupby('Ticker')['MACD'].mean())
```

```python
# RSI vs Daily Return
sns.scatterplot(data=df, x='RSI_14', y='Daily_Return_%')
plt.show()
```

```python
# MACD vs Signal Line
sns.scatterplot(data=df, x='MACD', y='Signal_Line')
plt.show()
```

```python
# Bollinger Bands
plt.figure(figsize=(12,5))
plt.plot(df['Close'], label='Close')
plt.plot(df['BB_Upper'], label='BB_Upper')
plt.plot(df['BB_Lower'], label='BB_Lower')
plt.legend()
plt.show()
```

```python
# SMA Analysis
plt.figure(figsize=(12,5))
plt.plot(df['Close'])
plt.plot(df['SMA_20'])
plt.plot(df['SMA_50'])
plt.legend(['Close','SMA20','SMA50'])
plt.show()
```

```python
# Company-wise feature boxplots
for col in ['RSI_14','MACD','Volume','Daily_Return_%']:
    plt.figure(figsize=(15,5))
    sns.boxplot(data=df, x='Ticker', y=col)
    plt.xticks(rotation=90)
    plt.show()
```
