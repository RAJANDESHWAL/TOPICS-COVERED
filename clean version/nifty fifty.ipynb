```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import FunctionTransformer, StandardScaler
import numpy as np

# 1. Strictly Positive Columns (Prices & Volume)
log_cols = [
    'Adj Close', 'Close', 'High', 'Low', 'Open', 'Volume',
    'SMA_20', 'SMA_50', 'EMA_12', 'EMA_26',
    'BB_Mid', 'BB_Upper', 'BB_Lower'
]

# 2. Oscillators & Returns (Naturally go negative or have bounds)
scale_cols = [
    'MACD', 'Signal_Line', 'Daily_Return_%', 'RSI_14'
]

# Pipeline for prices/volume
log_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('log', FunctionTransformer(np.log1p))
])

# Pipeline for oscillators/returns
scale_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Build the final preprocessor
preprocessor = ColumnTransformer([
    ('log_transforms', log_pipeline, log_cols),
    ('scale_transforms', scale_pipeline, scale_cols)
], remainder='passthrough') # 'passthrough' preserves any target variables or dates not listed above

# Execute the transformation safely
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

```python

log_cols = [
    'Adj Close', 'Close', 'High', 'Low', 'Open', 'Volume',
    'SMA_20', 'SMA_50', 'EMA_12', 'EMA_26',
    'BB_Mid', 'BB_Upper'
]

yeo_cols = [
    'MACD',
    'Signal_Line',
    'Daily_Return_%',
    'BB_Lower'
]


rsi_cols = ['RSI_14']


log_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('log', FunctionTransformer(np.log1p))
])


yeo_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('yeojohnson', PowerTransformer(method='yeo-johnson'))
])


rsi_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])


preprocessor = ColumnTransformer([
    ('log_transforms', log_pipeline, log_cols),
    ('yeo_transforms', yeo_pipeline, yeo_cols),
    ('rsi_transforms', rsi_pipeline, rsi_cols)
], remainder='passthrough')


X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

all_cols = log_cols + yeo_cols + rsi_cols

X_train_processed = pd.DataFrame(
    X_train_processed,
    columns=all_cols
)

X_test_processed = pd.DataFrame(
    X_test_processed,
    columns=all_cols
)
```

```python
## Hyperparamter tuning
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
penalty=['l1', 'l2', 'elasticnet']
c_values=[100,10,1.0,0.1,0.01]
solver=['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
class_weight=[{0:w,1:y} for w in [1,10,50,100] for y in [1,10,50,100]]
params=dict(penalty=penalty,C=c_values,solver=solver,class_weight=class_weight)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
cv=StratifiedKFold()
grid=GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv)
grid.fit(X_train_processed,y_train_encoded)
```
